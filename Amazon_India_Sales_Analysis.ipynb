## 6. Data Cleaning

Data cleaning is an important step in the data analysis process. It helps improve data quality by handling unnecessary columns, correcting data types, and identifying missing values.

The following operations will be performed:

- Remove unnecessary columns.
- Remove extra index columns.
- Convert numerical columns into appropriate data types.
- Convert date columns into datetime format.
- Standardize column names.
- Review missing values.

```python
# Create a copy of the original dataset
df_clean = df.copy()

# Remove unnecessary columns
df_clean.drop(columns=['index', 'Unnamed: 22'], inplace=True, errors='ignore')

# Remove extra spaces from column names
df_clean.columns = df_clean.columns.str.strip()

print("Unnecessary columns removed.")
print("Updated Dataset Shape:", df_clean.shape)

display(df_clean.head())
```

```python
# Convert Date column
df_clean['Date'] = pd.to_datetime(
    df_clean['Date'],
    errors='coerce'
)

# Convert numerical columns
df_clean['Amount'] = pd.to_numeric(
    df_clean['Amount'],
    errors='coerce'
)

df_clean['Qty'] = pd.to_numeric(
    df_clean['Qty'],
    errors='coerce'
)

print("Updated Data Types:")
print(df_clean.dtypes)
```

## 7. Missing Value Analysis

Missing values can affect the accuracy and reliability of data analysis.

In this section, we will examine missing values in each column and calculate their percentage.

We will not remove all missing records blindly. Appropriate handling methods will be selected according to the column and its importance.

```python
# Calculate missing values
missing_values = df_clean.isnull().sum()

# Calculate missing percentage
missing_percentage = (
    missing_values / len(df_clean)
) * 100

missing_summary = pd.DataFrame({
    'Missing Values': missing_values,
    'Missing Percentage': missing_percentage.round(2)
})

missing_summary = missing_summary.sort_values(
    by='Missing Values',
    ascending=False
)

display(missing_summary)
```

## 9. Feature Engineering

Feature engineering involves creating useful variables from existing dataset columns.

In this project, we will extract month, year, and day information from the order date to analyze sales patterns over time.

## 14. Machine Learning – Data Preparation

In this phase, we prepare the dataset for building a Machine Learning model.

The objective is to predict the sales amount using product and order-related information.

- **Target Variable:** Amount
- **Input Features:** Quantity, Category, Size, Fulfilment, Sales Channel, Shipping State, and date-related features.

Unnecessary identifier columns will be excluded from model training.

```python
# Select input features and target variable

features = [
    'Qty',
    'Category',
    'Size',
    'Fulfilment',
    'Sales Channel',
    'ship-service-level',
    'ship-state',
    'B2B',
    'Month',
    'Day'
]

target = 'Amount'

# Create model dataset
ml_data = df_clean[features + [target]].copy()

# Remove rows where target value is missing
ml_data = ml_data.dropna(subset=[target])

X = ml_data[features]
y = ml_data[target]

print("Machine Learning Dataset Prepared!")
print("Input Features Shape:", X.shape)
print("Target Shape:", y.shape)
print("Missing Target Values:", y.isnull().sum())
```

```python
from sklearn.model_selection import GroupShuffleSplit

# Keep all records from the same order in one split
groups = df_clean.loc[ml_data.index, 'Order ID']

splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx, test_idx = next(
    splitter.split(X, y, groups=groups)
)

X_train = X.iloc[train_idx]
X_test = X.iloc[test_idx]

y_train = y.iloc[train_idx]
y_test = y.iloc[test_idx]

print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)
```

## 15. Data Preprocessing and Model Building

Before training the model, missing values will be handled and categorical variables will be converted into numerical form using One-Hot Encoding.

A Random Forest Regressor will be used to predict the sales amount. This model combines multiple decision trees to generate predictions.

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestRegressor

# Separate numerical and categorical features
numeric_features = ['Qty', 'Month', 'Day']

categorical_features = [
    'Category',
    'Size',
    'Fulfilment',
    'Sales Channel',
    'ship-service-level',
    'ship-state',
    'B2B'
]

# Preprocessing for numerical data
numeric_transformer = SimpleImputer(strategy='median')

# Preprocessing for categorical data
categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])

# Combine preprocessing steps
preprocessor = ColumnTransformer(transformers=[
    ('num', numeric_transformer, numeric_features),
    ('cat', categorical_transformer, categorical_features)
])

# Create the complete model pipeline
model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('regressor', RandomForestRegressor(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    ))
])

# Train the model
model.fit(X_train, y_train)

print("Random Forest Model Trained Successfully!")
```

## 16. Model Evaluation

The trained model will be evaluated using the testing dataset.

The following evaluation metrics will be used:

- **MAE:** Measures the average absolute prediction error.
- **MSE:** Measures the average squared prediction error.
- **RMSE:** Measures prediction error in the original sales currency.
- **R² Score:** Measures how well the model explains variations in sales amount.

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# Generate predictions
y_pred = model.predict(X_test)

# Calculate evaluation metrics
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print("MODEL EVALUATION RESULTS")
print("-" * 35)
print("MAE:", round(mae, 2))
print("MSE:", round(mse, 2))
print("RMSE:", round(rmse, 2))
print("R2 Score:", round(r2, 4))
```

## 17. Actual vs Predicted Sales Analysis

This analysis compares the actual sales amounts with the amounts predicted by the Random Forest Regression model.

A scatter plot is used to visualize the prediction performance and identify differences between actual and predicted values.

```python
plt.figure(figsize=(10, 6))

plt.scatter(y_test, y_pred, alpha=0.4)

# Reference line
min_value = min(y_test.min(), y_pred.min())
max_value = max(y_test.max(), y_pred.max())

plt.plot(
    [min_value, max_value],
    [min_value, max_value],
    linestyle='--'
)

plt.title('Actual vs Predicted Sales Amount')
plt.xlabel('Actual Sales Amount (₹)')
plt.ylabel('Predicted Sales Amount (₹)')
plt.grid(True)
plt.show()
```

```python
comparison = pd.DataFrame({
    'Actual Amount': y_test.values,
    'Predicted Amount': y_pred
})

comparison['Difference'] = (
    comparison['Actual Amount'] -
    comparison['Predicted Amount']
)

display(comparison.head(10))
```

## 18. Feature Importance Analysis

Feature importance analysis identifies the input variables that contribute most to the Random Forest model's predictions.

It helps understand which product and order-related factors have a greater influence on the predicted sales amount.

```python
# Extract feature importance from the trained model

importance_values = model.named_steps['regressor'].feature_importances_

feature_names = model.named_steps['preprocessor'].get_feature_names_out()

feature_importance = pd.DataFrame({
    'Feature': feature_names,
    'Importance': importance_values
})

# Display top 15 important features
top_features = feature_importance.sort_values(
    by='Importance',
    ascending=False
).head(15)

plt.figure(figsize=(12, 7))

sns.barplot(
    data=top_features,
    x='Importance',
    y='Feature'
)

plt.title('Top 15 Important Features')
plt.xlabel('Importance Score')
plt.ylabel('Features')
plt.show()

display(top_features)
```
